# Combine

```python
import pikepdf

def merge_pdfs(input_pdfs, output_pdf):
    """
    input_pdfs: list of PDF file paths in desired order
    output_pdf: output merged PDF path
    """
    merged = pikepdf.Pdf.new()

    for pdf_path in input_pdfs:
        with pikepdf.open(pdf_path) as pdf:
            merged.pages.extend(pdf.pages)

    merged.save(output_pdf)

merge_pdfs(
    [
        "/content/Infy.pdf",
        "/content/Letter.pdf"
    ],
    "/content/Work.pdf"
)
```

```python
import subprocess

def compress_pdf(input_pdf, output_pdf, quality="/ebook"):
    """
    quality options:
    /screen  – lowest size, lower image quality
    /ebook   – good balance
    /printer – high quality
    /prepress – highest quality
    """
    subprocess.run([
        "gs",
        "-sDEVICE=pdfwrite",
        "-dCompatibilityLevel=1.4",
        f"-dPDFSETTINGS={quality}",
        "-dNOPAUSE",
        "-dQUIET",
        "-dBATCH",
        f"-sOutputFile={output_pdf}",
        input_pdf
    ], check=True)
```

```python
compress_pdf("Work.pdf", "Work2.pdf", quality = "/ebook")
```
